## Шаг 2. Инжиниринг признаков

### 2.1. Группы по срокам финансирования

Разделите все компании на три группы:

* Единичное финансирование — был всего один раунд финансирования.

* Срок финансирования до года — между первым и последним раундом финансирования прошло не более года.

* Срок финансирования более года.

Визуализируйте соотношение этих групп, создав два графика:

* По количеству компаний: Покажите, какой процент от общего числа компаний относится к каждой из трёх групп.
* По объёму инвестиций: Отобразите, какую долю от общего объёма привлечённых средств получила каждая группа.

```python
# Разделим все компании на три группы
def funding(row):
    if row['funding_rounds'] == 1:
        return 'Единичное финансирование'
    elif (row['last_funding_at'] - row['first_funding_at']).days <= 365:
        return 'Срок финансирования до года'
    else:
        return 'Срок финансирования более года'

df_investments['funding_group'] = df_investments.apply(funding, axis=1)
```

Определим компании с аномальным объёмом общего финансирования — используем метод IQR отдельно по каждому сегменту.

Определим сегменты рынка с наибольшей долей компаний, получивших аномальное финансирование, и выведим топ таких сегментов.

```python
# Группировка данных по столбцу market_class
grouped = df_investments.groupby('market_class')['funding_total_usd']

# Базовые статистики - всего, среднее значение, медиана
summary = grouped.agg(['count', 'mean', 'median']).round(0)
summary.columns = ['Всего', 'Среднее', 'Медиана']

# Подсчет выбросов для каждой группы
outliers = []
for segment in summary.index:
    data = df_investments[df_investments['market_class'] == segment]['funding_total_usd']
    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)
    IQR = Q3 - Q1
    upper = Q3 + 1.5 * IQR
    outliers.append(len(data[data > upper]))

summary['Выбросы'] = outliers
summary['Доля_%'] = (summary['Выбросы'] / summary['Всего'] * 100).round(1)

# Вывод результата
print("Сегменты с наибольшей долей выбросов:")
print(summary.sort_values('Доля_%', ascending=False).head(10))
```

Исключим из датасета компании, которые ранее посчитали получившими аномальное финансирование.

```python
# Пересчитываем выбросы с использованием оригинального столбца market
segments_to_remove = []

# Для каждого сегмента в столбце market
for segment in df_investments['market'].unique():
    # Берем данные только для этого сегмента
    segment_data = df_investments[df_investments['market'] == segment]['funding_total_usd']
    
    # Проверяем, что в сегменте достаточно данных для анализа
    if len(segment_data) > 10:
        Q1 = segment_data.quantile(0.25)
        Q3 = segment_data.quantile(0.75)
        IQR = Q3 - Q1
        
        if IQR > 0:
            upper = Q3 + 1.5 * IQR
            
            # Находим выбросы в этом сегменте
            outliers = df_investments[
                (df_investments['market'] == segment) & 
                (df_investments['funding_total_usd'] > upper)
            ]
            segments_to_remove.extend(outliers.index)

# Удаляем выбросы
before_count = len(df_investments)
df_investments = df_investments.drop(segments_to_remove)
after_count = len(df_investments)

print(f"Статистика удаления выбросов:")
print(f"• Было компаний: {before_count}")
print(f"• Удалено выбросов: {len(segments_to_remove)}")
print(f"• Осталось компаний: {after_count}")
print(f"• Процент удаленных: {len(segments_to_remove)/before_count*100:.1f}%")

# Пересчитываем классификацию сегментов
market_count = df_investments['market'].value_counts()
mass_market = market_count[market_count > 120].index.tolist()
mid_market = market_count[(market_count >= 35) & (market_count <= 120)].index.tolist()

def class_market(market):
    if market in mass_market:
        return market
    elif market in mid_market:
        return 'mid'
    else:
        return 'niche'

# Обновляем столбец market_class
df_investments['market_class'] = df_investments['market'].apply(class_market)

print(f"\nОбновленная статистика после удаления выбросов:")
print(f"• Уникальных сегментов (market): {df_investments['market'].nunique()}")
print(f"• Уникальных значений (market_class): {df_investments['market_class'].nunique()}")
print(f"• Массовых сегментов: {len(mass_market)}")
print("\nТоп-10 сегментов после очистки:")
print(df_investments['market_class'].value_counts().head(10))
```

```python
# Создаем DataFrame для анализа выбросов по сегментам
outliers_by_segment = []

# Для каждого сегмента в столбце market
for segment in df_investments['market'].unique():
    # Берем данные только для этого сегмента
    segment_data = df_investments[df_investments['market'] == segment]['funding_total_usd']
    
    # Проверяем, что в сегменте достаточно данных для анализа
    if len(segment_data) > 10:
        Q1 = segment_data.quantile(0.25)
        Q3 = segment_data.quantile(0.75)
        IQR = Q3 - Q1
        
        if IQR > 0:
            upper = Q3 + 1.5 * IQR
            
            # Находим выбросы в этом сегменте
            outliers_count = len(df_investments[
                (df_investments['market'] == segment) & 
                (df_investments['funding_total_usd'] > upper)
            ])
            
            total_in_segment = len(segment_data)
            outlier_percentage = (outliers_count / total_in_segment * 100) if total_in_segment > 0 else 0
            
            outliers_by_segment.append({
                'Сегмент': segment,
                'Всего компаний': total_in_segment,
                'Выбросы': outliers_count,
                'Доля выбросов, %': round(outlier_percentage, 1)
            })

# Создаем DataFrame и сортируем по доле выбросов
outliers_df = pd.DataFrame(outliers_by_segment)
outliers_df = outliers_df.sort_values('Доля выбросов, %', ascending=False)

print("Топ-10 сегментов по доле выбросов (аномального финансирования):")
print(outliers_df[['Сегмент', 'Всего компаний', 'Выбросы', 'Доля выбросов, %']].head(10).to_string(index=False))
```

После исключения аномальных записей, на основе столбцов `mid_funding_at` и `funding_rounds` оставим в датасете данные только об определённых компаниях. Они должны были получать финансирование в годы, когда было зафиксировано 50 или более раундов финансирования.

```python
# Годы финансирования
df_investments['year'] = df_investments['mid_funding_at'].dt.year

# Считаем раунды по годам
rounds = df_investments.groupby('year')['funding_rounds'].count()

# Находим первый год с 50+ раундами
first_good_year = rounds[rounds >= 50].index.min()
print(f"Первый год с 50+ раундами: {first_good_year}")

# Берем все годы начиная с первого хорошего года
df_investments = df_investments[df_investments['year'] >= first_good_year]

print(f"Осталось компаний: {len(df_investments)}")
print(f"Период анализа: {df_investments['year'].min()} - {df_investments['year'].max()} годы")
```

На основе графика сделаем вывод о том, какие сегменты показывают наиболее быстрый и уверенный рост.

```python
# 1. Создаем сводную таблицу
pivot_table = pd.pivot_table(
    df_investments,
    values='funding_total_usd',
    index='year',
    columns='market_class',
    aggfunc='sum'
)

print("Сводная таблица финансирования по годам и сегментам:")
pivot_table
```

Отберём из неё только те сегменты, которые показывали рост размера суммарного финансирования в 2014 году по сравнению с 2013.

```python
# Находим сегменты с ростом в 2014
growth_segments = []

for segment in pivot_table.columns:
    if 2013 in pivot_table.index and 2014 in pivot_table.index:
        funding_2013 = pivot_table.loc[2013, segment]
        funding_2014 = pivot_table.loc[2014, segment]
        
        if pd.notna(funding_2013) and pd.notna(funding_2014):
            if funding_2014 > funding_2013:
                growth = ((funding_2014 - funding_2013) / funding_2013 * 100)
                growth_segments.append({
                    'Сегмент': segment,
                    '2013': int(funding_2013),
                    '2014': int(funding_2014),
                    'Рост_%': round(growth, 1)
                })

# Создаем и сортируем
growth_df = pd.DataFrame(growth_segments).sort_values('Рост_%', ascending=False)

print(f"Сегменты с ростом финансирования в 2014 (всего {len(growth_df)}):")
print(growth_df.to_string(index=False))
```

### 4.3 Годовая динамика доли возвращённых средств по типам финансирования

Заказчик хочет знать, какая часть вложенных или выданных денег со временем возвращается обратно инвесторам или финансистам. Ваша цель — для каждого года и каждого вида финансирования рассчитать нормированные значения возврата средств: то есть какую долю возвращённые средства составляют от предоставленных. При этом слишком большие аномальные значения, то есть неадекватные выбросы, нужно заменить на пропуски.

Совет: когда будете делить сумму возвращённых средств на суммарный объём привлечённого финансирования по конкретному году, добавьте к знаменателю небольшое число, например `1e-60`. Это поможет избежать деления на ноль.

```python
# Сумма инвестиций по годам для каждого типа финансирования
invest_by_year = df_investments.groupby('year')[finance_cols].sum()

# доля возвратов
return_percent = df_returns.copy()

for col in finance_cols:
    if col in df_returns.columns:
        return_percent[col] = (df_returns[col] * 1e6) / (invest_by_year[col] + 1e-60) * 100

# Убираем аномалии (>100%) и округляем
for col in finance_cols:
    if col in return_percent.columns:
        return_percent.loc[return_percent[col] > 100, col] = np.nan
        return_percent[col] = return_percent[col].round(2)

print("Доля возвратов в %:")
print(return_percent)
```

Постройте график, на котором отобразите нормированные значения возврата средств для типов финансирования `venture`, `debt_financing`, `private_equity`, `seed` и `angel`.

Сделайте вывод о том, в каких типах финансирования наблюдается наиболее устойчивый рост показателя.

```python
# Выбираем нужные типы финансирования
selected_types = ['venture', 'debt_financing', 'private_equity', 'seed', 'angel']

# График
plt.figure(figsize=(12, 6))

for col in selected_types:
    if col in return_percent.columns:
        plt.plot(return_percent.index, return_percent[col], marker='o', label=col, linewidth=2)

plt.title('Доля возвращённых средств по типам финансирования (%)')
plt.xlabel('Год')
plt.ylabel('Доля возврата, %')
plt.legend()
plt.grid(True, alpha=0.3)
plt.xticks(return_percent.index)
plt.tight_layout()
plt.show()
```

- Наиболее устойчивый рост: Venture (венчурные): с 2011 по 2014 год стабильный рост с 25.42% до 68.13%, показывают уверенный рост после 2011 года.

- Переменные, но с потенциалом: Seed (посевные): снижение с 2006 (93.93%) до 2013 (14.79%), восстановление в 2014 до 27.68%, высокая волатильность. Debt Financing (долговое): нестабильные показатели 22-96%, в 2014 году - 96.20% (максимальное значение). Private Equity (частные): высокая волатильность от 8.00% до 93.55%.

- Нет четкого тренда: Angel (бизнес-ангелы): циклический характер 9-84%, данные за 2014 отсутствуют.
